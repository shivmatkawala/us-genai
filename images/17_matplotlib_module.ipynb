##### Matplotlib

Matplotlib is the most widely used plotting library
in python for creating static, animated and interactive
visualizations.

It is especially popular in data science, machine learning,
scientific computing and software development.

It provides fine-grained control over every aspect of a plot.


Why use matplotlib ?
    # Create line bar, scatter, pie, histogram and many other charts.

    It works seamlessly with numpy and pandas.

    can save plots in PNG, PDF, SVG, JPEG formats.


Installation:
    # python -m pip install matplotlib

```python
# IMPORT MATPLOTLIB
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

plt.plot(x, y)
plt.show()
```

```python
# Line Plot:
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [540, 330, 700, 450, 40]

plt.plot(months, sales, 
         color='red', 
         linestyle="--", 
         marker='*', 
         linewidth=2, 
         markersize=10,
         )

plt.xlabel("Month")
plt.ylabel("Sale")

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
mango_sale = [400, 500, 350, 600, 750]
grapes_sale = [200, 740, 400, 300, 100]

plt.plot(months, mango_sale,
        label="Mango Sale",
        color='red', 
         linestyle="--", 
         marker='*', 
         linewidth=2, 
         markersize=10,
        )
plt.plot(months, grapes_sale, 
         label="Grapes Sale",
         color='blue', 
         linestyle="--", 
         marker='o', 
         linewidth=2, 
         markersize=10,
         )

plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Scatter Plot:

height = [150, 160, 170, 180, 190]
weight = [50, 60, 70, 80, 90]

plt.scatter(height, weight, marker="*", s=200, alpha=0.2)

plt.xlabel("Height")
plt.ylabel("Weight")

plt.title("Height Vs Weight")

plt.show()
```
