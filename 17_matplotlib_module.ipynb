##### Matplotlib

Matplotlib is the most widely used plotting library
in python for creating static, animated and interactive
visualizations.

It is especially popular in data science, machine learning,
scientific computing and software development.

It provides fine-grained control over every aspect of a plot.


Why use matplotlib ?
    # Create line bar, scatter, pie, histogram and many other charts.

    It works seamlessly with numpy and pandas.

    can save plots in PNG, PDF, SVG, JPEG formats.


Installation:
    # python -m pip install matplotlib

```python
# IMPORT MATPLOTLIB
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 30, 40]

plt.plot(x, y)
plt.show()
```

```python
# Line Plot:
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [540, 330, 700, 450, 40]

plt.plot(months, sales, 
         color='red', 
         linestyle="--", 
         marker='*', 
         linewidth=2, 
         markersize=10,
         )

plt.xlabel("Month")
plt.ylabel("Sale")

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
mango_sale = [400, 500, 350, 600, 750]
grapes_sale = [200, 740, 400, 300, 100]

plt.plot(months, mango_sale,
        label="Mango Sale",
        color='red', 
         linestyle="--", 
         marker='*', 
         linewidth=2, 
         markersize=10,
        )
plt.plot(months, grapes_sale, 
         label="Grapes Sale",
         color='blue', 
         linestyle="--", 
         marker='o', 
         linewidth=2, 
         markersize=10,
         )

plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Scatter Plot:

height = [150, 160, 170, 180, 190]
weight = [50, 60, 70, 80, 90]

plt.scatter(height, weight, marker="*", s=200, alpha=0.2)

plt.xlabel("Height")
plt.ylabel("Weight")

plt.title("Height Vs Weight")

plt.show()
```

Area Chart:

```python
days = [1, 2, 3, 4, 5]
visitors = [100, 120, 150, 130, 180]

plt.fill_between(days, visitors)

plt.title("Website Visitors")
plt.xlabel("Day")
plt.ylabel("Visitors")
plt.show()
```

Box Plot

```python
salary = [25, 30, 35, 40, 45, 50, 55, 60, 100]

plt.boxplot(salary)
plt.title("Salary Distribution")
plt.show()
```

Histogram:

```python
ages = [18, 19, 20, 20, 21, 22, 22, 22, 23, 24, 25, 25, 26, 27, 28, 28, 29, 30]

plt.hist(ages, bins=6, edgecolor="red")

plt.title("Age Distribution")
plt.xlabel("age")
plt.ylabel("Frquency")

plt.show()
```

Pie Chart:

```python
department = ["IT", "HR", "Sales", "Marketing"]
employees = [40, 15, 25, 20]

plt.pie(employees, labels=department, autopct='%1.1f%%')
plt.title("Department wise employees")


plt.show()
```

Horizontal Bar Chart

```python
languages = ["Python", "Java", "C++", "JavaScript"]
users = [90, 75, 50, 85]

plt.barh(languages, users)
plt.title("Popularity Score")
plt.show()
```
